```python
from IPython.core.display import warnings
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset("titanic")
```

```python
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.drop(["alive", "embark_town","adult_male","who","class","deck"], axis = 1, inplace=True)
```

```python
df['age'].fillna(df['age'].median(), inplace=True)
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
df.head()
```

```python
df['sex'] = le.fit_transform(df['sex'])
```

```python
df['embarked'] = le.fit_transform(df['embarked'])
```

```python
df.head()
```

```python
df['alone'] = le.fit_transform(df['alone'])
```

```python
df.head()
```

```python
df = df.astype(int)
```

```python
df.head()
```

```python
x = df.drop("survived", axis=1)
y = df["survived"]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

```

# **Logistic Regression Model**

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()

model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
accuracy_score(y_test, y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```

```python
from sklearn.preprocessing import StandardScaler
scalar = StandardScaler()
```

```python
x_train_scaled = scalar.fit_transform(x_train)
x_train_scaled
```

```python
x_test_scaled = scalar.transform(x_test)
x_test_scaled
```

# KNN **Model**

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train_scaled, y_train)
```

```python
y_pred_knn = knn.predict(x_test_scaled)
```

```python
accuracy_score(y_test, y_pred_knn)
```

```python
confusion_matrix(y_test, y_pred_knn)
```

```python
print(classification_report(y_test, y_pred_knn))
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

y_pred_lr  = model.predict(x_test)
y_pred_knn = knn.predict(x_test_scaled)

results = pd.DataFrame({
    "Model": ["Logistic Regression", "KNN"],
    "Accuracy":  [accuracy_score(y_test, y_pred_lr),  accuracy_score(y_test, y_pred_knn)],
    "Precision": [precision_score(y_test, y_pred_lr), precision_score(y_test, y_pred_knn)],
    "Recall":    [recall_score(y_test, y_pred_lr),    recall_score(y_test, y_pred_knn)],
    "F1-score":  [f1_score(y_test, y_pred_lr),        f1_score(y_test, y_pred_knn)],
}).round(4)

results
```

```python
results.set_index("Model").plot(kind="bar", figsize=(8,5))
plt.title("Logistic Regression vs KNN")
plt.ylabel("Score")
plt.ylim(0.6, 1.0)
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.tight_layout()
plt.savefig("model_comparison.png", dpi=150)
plt.show()
```
